```python
%pip install psycopg2-binary sqlalchemy pandas matplotlib -q

import sqlalchemy
import pandas as pd
import matplotlib.pyplot as plt

engine = sqlalchemy.create_engine('postgresql://joomlauser:joomlapassword@database:5432/joomladb')
print("Conexión exitosa a PostgreSQL 16")
```

```python
df_tablas = pd.read_sql("SELECT table_name FROM information_schema.tables WHERE table_schema = 'public';", engine)
print(f"Total de tablas encontradas en joomladb: {len(df_tablas)}")
df_tablas.head(10)
```

```python
plt.figure(figsize=(8, 4))
df_tablas['prefijo'] = df_tablas['table_name'].apply(lambda x: x.split('_')[0])
conteo_prefijos = df_tablas['prefijo'].value_counts()

conteo_prefijos.plot(kind='bar', color='skyblue')
plt.title('Distribución de Tablas por Prefijo en PostgreSQL')
plt.xlabel('Prefijo de Tabla')
plt.ylabel('Cantidad de Tablas')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
